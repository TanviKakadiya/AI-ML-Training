```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
classes = ['Python','R','AI','ML','DS']
class1_students=[30,10,20,25,10]
class2_students=[40,5,20,20,10]
class3_students=[35,5,30,15,15]
```

```python
plt.bar(classes,class1_students)
```

```python
plt.bar(classes,class2_students)
```

```python
plt.bar(classes,class3_students)
```

```python
plt.barh(classes,class1_students)
```

```python
plt.bar(classes,class1_students,width=0.2,align='edge',edgecolor='r',color='pink',linewidth=5,alpha = 0.5)
```

```python
plt.bar(classes,class1_students,width=0.5,align='edge',edgecolor='purple',color='pink',linewidth=2,alpha = 0.5,linestyle='--')
```

```python
plt.bar(classes,class1_students,width=0.5,align='edge',edgecolor='purple',color='pink',linewidth=2,alpha = 0.5,linestyle='--',label = ' class 1 students', visible = False)
```

```python
plt.bar(classes,class1_students,width=0.5,align='edge',edgecolor='purple',color='pink',linewidth=2,alpha = 0.5,linestyle='--',label = ' class 1 students')
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
plt.barh(classes,class1_students,align='edge',edgecolor='purple',color='pink',linewidth=2,alpha = 0.5,linestyle='--',label = ' class 1 students')
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
plt.bar(classes,class1_students,align='edge',edgecolor='purple',color='pink',linewidth=2,alpha = 0.5,linestyle='--',label = ' class 1 students')

plt.title('Bar chart of MU class',fontsize = 15)
plt.xlabel('classes',fontsize=20)
plt.ylabel('no.of students',fontsize=20)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(16,9))
plt.bar(classes,class1_students,width=0.3,color='pink',label = ' class 1 students')
plt.bar(classes,class2_students,width=0.5,color='m',label='class 2 students')
plt.title('Bar chart of MU class',fontsize = 15)
plt.xlabel('classes',fontsize=20)
plt.ylabel('no.of students',fontsize=20)
plt.show()
```

```python
import numpy as np

style.use('ggplot')
plt.figure(figsize=(16,9))
classes_index=np.arange(len(classes))
width=0.2
plt.bar(classes_index,class1_students,width=0.6,color='pink',label = ' class 1 students')
plt.bar(classes_index+width,class2_students,width=0.6,color='deeppink',label='class 2 students')
plt.xticks(classes_index+width,classes,rotation=20)#x label manipulation
plt.title('Bar chart of MU class',fontsize = 15)
plt.xlabel('classes',fontsize=20)
plt.ylabel('no.of students',fontsize=20)
plt.show()
```

```python
import numpy as np

style.use('ggplot')
plt.figure(figsize=(16,9))
classes_index=np.arange(len(classes))
width=0.2
plt.bar(classes_index,class1_students,width=0.6,color='deeppink',label = ' class 1 students')
plt.bar(classes_index+width,class2_students,width=0.6,color='pink',label='class 2 students')

plt.bar(classes_index+width+width,class3_students,width=0.6,color='hotpink',label='class 2 students')
plt.xticks(classes_index+width,classes,rotation=20)#x label manipulation
plt.title('Bar chart of MU class',fontsize = 15)
plt.xlabel('classes',fontsize=20)
plt.ylabel('no.of students',fontsize=20)
plt.show()
```
