```python
sns.lineplot(x='size',y="total_bill",data=tips_df,hue='sex',style='sex',palette = "rocket",dashes=False,markers=['o','<'],legend='brief')
```

```python
sns.lineplot(x='size',y="total_bill",data=tips_df,hue='day',style='day',palette = "rocket",dashes=False,markers=['o','<','>','^'],legend='brief')

plt.title('Line plot',fontsize=19)
plt.xlabel('Size',fontsize=15)
plt.ylabel('Total Bill',fontsize = 15)
plt.show()
```
