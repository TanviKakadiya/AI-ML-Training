```python
ml_students_age=np.random.randint(18,50,(100))
py_students_age=np.random.randint(14,50,(100))

bins = [15,20,25,30,35,40,45]

plt.hist(ml_students_age,bins)
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.show()
```

```python
ml_students_age=np.random.randint(18,50,(100))
py_students_age=np.random.randint(14,50,(100))

bins = [15,20,25,30,35,40,45]

plt.hist(ml_students_age,bins,rwidth=0.90)
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth=0.90,histtype='step')
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth=0.90,histtype='barstacked')
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth=0.90,histtype='barstacked',align='left')
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth=0.90,histtype='barstacked',align='left',orientation='horizontal')
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth=0.90,histtype='barstacked',align='left',color='pink',label='Ml Student')
plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.legend()
plt.show()
```

```python
bins = [15,20,25,30,35,40,45]

plt.hist(ml_students_age,bins,rwidth=0.8,histtype='bar',align='left',orientation='horizontal',color='pink',label='Ml Student')

plt.hist(py_students_age,bins,rwidth=0.80,histtype='barstacked',align='left',color='hotpink',label='Py Student')


plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.legend()
plt.show()
```

```python
bins = [15,20,25,30,35,40,45]

plt.hist([ml_students_age,py_students_age],bins,rwidth=0.8,histtype='bar',align='left',color=['pink','hotpink'],label=['Ml Student','PY Student'])


plt.title('Ml students age Histograms')
plt.xlabel('Students age category')
plt.ylabel('no. of students age')
plt.legend()
plt.show()
```
